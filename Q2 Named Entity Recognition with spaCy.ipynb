```python
# Q2: Named Entity Recognition with spaCy
import spacy
nlp = spacy.load("en_core_web_sm")

text = "Barack Obama served as the 44th President of the United States and won the Nobel Peace Prize in 2009."
doc = nlp(text)
for ent in doc.ents:
    print(f"Text: {ent.text}, Label: {ent.label_}, Start: {ent.start_char}, End: {ent.end_char}")

# --- Short Answer ---
print("\nQ2 Short Answers:")
print("1. NER vs POS Tagging:")
print("   - NER identifies named entities (people, dates, etc.).")
print("   - POS tagging assigns parts of speech (noun, verb, etc.) to each word.")

print("2. Applications of NER:")
print("   - Extracting company names from financial news.")
print("   - Recognizing locations or dates in search engine queries.")
```
