```python
# Q4: Sentiment Analysis using HuggingFace Transformers
from transformers import pipeline


classifier = pipeline("sentiment-analysis", model="distilbert/distilbert-base-uncased-finetuned-sst-2-english")
result = classifier("Despite the high price, the performance of the new MacBook is outstanding.")[0]
print(f"Sentiment: {result['label']}")
print(f"Confidence Score: {result['score']:.4f}")

# --- Short Answer ---
print("\nQ4 Short Answers:")
print("1. Difference between BERT and GPT:")
print("   - BERT uses only encoder blocks.")
print("   - GPT uses only decoder blocks.")

print("2. Benefit of pre-trained models:")
print("   - Saves time and resources.")
print("   - Leverages large-scale training for better accuracy.")
```
